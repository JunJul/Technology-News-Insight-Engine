```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
```

```python
cleaned_df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Technology News Insight Engine/Data/Hackernoon_cleaned_subsets.csv')
uncleaned_df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Technology News Insight Engine/Data/Hackernoon_uncleaned_all_subsets.csv')
```

```python
uncleaned_docs = uncleaned_df['article'].astype(str)
cleaned_docs = cleaned_df['lemmatized_article'].astype(str)
```

```python
uncleaned_words = [len(doc.split()) for doc in uncleaned_docs]
cleaned_words = [len(doc.split()) for doc in cleaned_docs]
```
